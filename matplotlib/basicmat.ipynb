```python
a = [1,2,3,4]
label = ["burger","pizza","fries","noodles"]
plt.pie(a,labels=label)
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","fries","noodles"]
plt.pie(a,labels=label,explode=[0,0.2,0,0])
plt.show()
# explode se jaha hume explode karna h kar dege
```

```python
a = [1,2,3,4]
label = ["burger","pizza","fries","noodles"]
plt.pie(a,labels=label,explode=[0,0.2,0,0],shadow=True)
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","fries","noodles"]
color = ['pink','brown','yellow','grey']
plt.pie(a,labels=label,explode=[0,0.2,0,0],shadow=True,colors = c)
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","fries","noodles"]
color = ['pink','brown','yellow','grey']
plt.pie(a,labels=label,explode=[0,0.2,0,0],shadow=True,colors = c,autopct='%1.1f%%')
plt.show()
# autopct show the percent of every part 
```

```python
a = [1,2,3,4]
label = ["burger","pizza","fries","noodles"]
color = ['pink','brown','yellow','grey']
plt.pie(a,labels=label,explode=[0,0.2,0,0],shadow=True,colors = c,autopct='%1.1f%%')
plt.legend()
plt.show()
# legend show the status of the pie chart
```
