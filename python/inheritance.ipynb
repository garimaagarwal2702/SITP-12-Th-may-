## Inheritance in Python

Inheritance is an Object-Oriented Programming (OOP) concept where one class acquires the properties and methods of another class.

The class whose properties are inherited is called the Parent Class / Base Class / Super Class
The class that inherits is called the Child Class / Derived Class / Sub Class

**Inheritance helps in:**
* Code reusability
* Reducing duplication
* Better organization of code
* Easy maintenance

Types of Inheritance in Python

There are mainly 5 types of inheritance:
* Single Inheritance
* Multiple Inheritance
* Multilevel Inheritance
* Hierarchical Inheritance
* Hybrid Inheritance

### 1. Single Inheritance

One child class inherits from one parent class.

### 2. Multiple Inheritance

One child class inherits from more than one parent class.

### 3. Multilevel Inheritance

A class inherits from a class that already inherited another class.

### 4. Hierarchical Inheritance

Multiple child classes inherit from the same parent class.

### 5. Hybrid Inheritance

Combination of two or more types of inheritance.


```python
# Single Inheritance

class College:
    college_name = "SGI"
    college_address = "Sikar"
    collegesp = 43 # ek class se method ke ander properties inherit karne ke liye self is used
    def show(self):
        print(self.college_name)
        print(self.college_address)
        print(self.collegesp)

class Student(College):
    student_name = "Garima"
    student_rollnumber = 19
    student_number = 12345678
    student_branch = "CSE"
st = Student()
st.show()
print(st.student_branch)
```

```python
# multilevel Inheritance
class person:
    pass
class teacher(person):
    pass
class HOD(teacher):
    pass 
```

```python
class Person:
    def person_info(self):
        print("i am a person")
class Teacher(Person):
    def teacher_info(self):
        print("i am a teacher")
class HOD(Teacher):
    def hod_info(self):
        print("i am HOD")

h = HOD()
h.person_info()
h.teacher_info()
h.hod_info()
```

```python
# multiple inheritance

class Developer:
    def coding(self):
        print("write the code")
class Manager:
    def meeting(self):
        print("He don't know anything")
class Team_Leader(Developer, Manager):
    def project(self):
        print("Handling project")
    
team = Team_Leader()
team.coding()
team.project()
team.meeting()
```

```python
# Company system:- two companies give their data to one another company which can acess both of their data

class Company1:
    sells = 20
    in_days = 1
    income = 20000
    def company1_info(self):
        print("sells of company:-",self.sells)
        print("sells in days:-",self.in_days)
        print("income :-",self.income)
class Company2:
    sells1 = 30
    in_days1 = 2
    income1 = 30000
    def company2_info(self):
        print("sells of company:-",self.sells1)
        print("sells in days:-",self.in_days1)
        print("income:-",self.income1)
class Company3(Company1,Company2):
    sells2 = 50
    in_days2 = 2
    income2 = 50000
    def company3_info(self):
        print("sells of company:-",self.sells2)
        print("sells in days:-",self.in_days2)
        print("income:-",self.income2)

company = Company3()
company.company1_info()
```

```python
##Hierachial Inheritance

class Developer:
    def coding(self):
        print("I do coding")
class Manager(Developer):
    def meeting(self):
        print("i just do meeting")
class Team_leader(Developer):
    def project():
        print("handling Project")
class Designer(Developer):
    def designing(self):
        print("i do designing")

man = Manager()
Team = Team_leader()
design = Designer()
man.coding()
Team.coding()
design.coding()
```

```python
## Hybrid Inheritance

class son:
    def Son(self):
        print("i am son")
class daughter:
    def Daughter(self):
        print("i am daughter")
class mother(son,daughter):
    def Mother(self):
        print("i am mother")
class father(mother):
    def Father(self):
        print("I am father")
mom = mother()
dad = father()
mom.Daughter()
dad.Son()
```

```python
# 1. Create a class Employee with attributes emp_name, emp_id, and salary. 
# Create 3 objects and display their details
class Employee:
    def __init__(self, emp_name, emp_id, salary):
        self.emp_name = emp_name
        self.emp_id = emp_id
        self.salary = salary

    def display(self):
        print("Employee Name:", self.emp_name)
        print("Employee ID:", self.emp_id)
        print("Salary:", self.salary)

# Creating 3 objects
emp1 = Employee("Garima", 101, 50000)
emp2 = Employee("Hemant", 102, 60000)
emp3 = Employee("Priya", 103, 55000)

# Displaying details
emp1.display()
emp2.display()
emp3.display()
```

```python
# 2. Create a class Calculator with methods for addition, subtraction, 
#multiplication, and division. 

class Calculator:
    def add(self):
        a = float(input("Enter a number to add "))
        b = float(input("input another number to add"))
        self.c = a + b
        print("sum=",self.c)
    def sub(self):
        d = float(input("Enter your number to sub"))
        e = float(input("Enter another number to sub"))
        self.f = d - e
        print("sub=",self.f)
    def mul(self):
        g = float(input("Enter a number to multiply"))
        h = float(input("Enter another number to multiply"))
        self.i = g * h
        print("Multiplication=",self.i)
    def devide(self):
        j = float(input("Enter a number to devision"))
        k = float(input("Enter another number to devision"))
        self.l = j / k
        print("Devide = ",self.l)
cal = Calculator()  
cal.add()
cal.sub()
cal.mul()
cal.devide()
```

```python
# 3. Create a class Book with a constructor to initialize title, author, and price, 
# then print details.
class Book:
    def __init__(self, title, author, price ):
        self.title = title 
        self.author = author 
        self.price = price 

    def display(self):
        print("Title of Book:", self.title)
        print("Author of Book:", self.author)
        print("Price of Book:", self.price)

# Creating wanted objects
book1 = Book("Programming","Garima Agarwal",1500)
book2 = Book("India GK","Hemant Agarwal",800)
book3 = Book("News Analysis","Priya Agarwal",1000)

# Displaying details
book1.display()
book2.display()
book3.display()
```

```python
# 4. Create a class Vehicle with method display(). Inherit it into class Car and 
#call parent method. 
class Vehicle:
    def display(self):
        print("This is a vehicle")
        print("It is very affordable")
class Car(Vehicle):
    def show(self):
        # Calling parent class method
        self.display()
        print("This is a car")


# Create object of Car
c = Car()

# Call method
c.show()
```
