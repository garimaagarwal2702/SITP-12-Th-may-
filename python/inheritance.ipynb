## Inheritance in Python

Inheritance is an Object-Oriented Programming (OOP) concept where one class acquires the properties and methods of another class.

The class whose properties are inherited is called the Parent Class / Base Class / Super Class
The class that inherits is called the Child Class / Derived Class / Sub Class

**Inheritance helps in:**
* Code reusability
* Reducing duplication
* Better organization of code
* Easy maintenance

Types of Inheritance in Python

There are mainly 5 types of inheritance:
* Single Inheritance
* Multiple Inheritance
* Multilevel Inheritance
* Hierarchical Inheritance
* Hybrid Inheritance

### 1. Single Inheritance

One child class inherits from one parent class.

### 2. Multiple Inheritance

One child class inherits from more than one parent class.

### 3. Multilevel Inheritance

A class inherits from a class that already inherited another class.

### 4. Hierarchical Inheritance

Multiple child classes inherit from the same parent class.

### 5. Hybrid Inheritance

Combination of two or more types of inheritance.


```python
# Single Inheritance

class College:
    college_name = "SGI"
    college_address = "Sikar"
    collegesp = 43 # ek class se method ke ander properties inherit karne ke liye self is used
    def show(self):
        print(self.college_name)
        print(self.college_address)
        print(self.collegesp)

class Student(College):
    student_name = "Garima"
    student_rollnumber = 19
    student_number = 12345678
    student_branch = "CSE"
st = Student()
st.show()
print(st.student_branch)
```

```python
# multilevel Inheritance
class person:
    pass
class teacher(person):
    pass
class HOD(teacher):
    pass 
```

```python
class Person:
    def person_info(self):
        print("i am a person")
class Teacher(Person):
    def teacher_info(self):
        print("i am a teacher")
class HOD(Teacher):
    def hod_info(self):
        print("i am HOD")

h = HOD()
h.person_info()
h.teacher_info()
h.hod_info()
```

```python
# multiple inheritance

class Developer:
    def coding(self):
        print("write the code")
class Manager:
    def meeting(self):
        print("He don't know anything")
class Team_Leader(Developer, Manager):
    def project(self):
        print("Handling project")
    
team = Team_Leader()
team.coding()
team.project()
team.meeting()
```

```python
# Company system:- two companies give their data to one another company which can acess both of their data

class Company1:
    sells = 20
    in_days = 1
    income = 20000
    def company1_info(self):
        print("sells of company:-",self.sells)
        print("sells in days:-",self.in_days)
        print("income :-",self.income)
class Company2:
    sells1 = 30
    in_days1 = 2
    income1 = 30000
    def company2_info(self):
        print("sells of company:-",self.sells1)
        print("sells in days:-",self.in_days1)
        print("income:-",self.income1)
class Company3(Company1,Company2):
    sells2 = 50
    in_days2 = 2
    income2 = 50000
    def company3_info(self):
        print("sells of company:-",self.sells2)
        print("sells in days:-",self.in_days2)
        print("income:-",self.income2)

company = Company3()
company.company1_info()
```

```python
##Hierachial Inheritance

class Developer:
    def coding(self):
        print("I do coding")
class Manager(Developer):
    def meeting(self):
        print("i just do meeting")
class Team_leader(Developer):
    def project():
        print("handling Project")
class Designer(Developer):
    def designing(self):
        print("i do designing")

man = Manager()
Team = Team_leader()
design = Designer()
man.coding()
Team.coding()
design.coding()
```

```python
## Hybrid Inheritance

class son:
    def Son(self):
        print("i am son")
class daughter:
    def Daughter(self):
        print("i am daughter")
class mother(son,daughter):
    def Mother(self):
        print("i am mother")
class father(mother):
    def Father(self):
        print("I am father")
mom = mother()
dad = father()
mom.Daughter()
dad.Son()
```
