### Constructor
A constructor in Python is a special method that is automatically called when an object of a class is created.

It is mainly used to:
* Initialize object data
* Assign values to variables
* Perform setup work

```python
class college:
    def __init__(self): # this is the syntax of constructor
        pass
clg = college() 
##__init__ is called automatically when an object is created.
# It always starts and ends with double underscores.
```

```python
class college:
    def __init__(self):
        college_name = "SGI"
clg.college()
clg.college_name()
# here will be no attribute with college_name because in constructor
# we use self. before the vaiable name in function to define variable.
```

```python
class college:
    def __init__(self):
        self.college_name = "SGI" # instance variable
        self.college_address = "Sikar" # instance variable
        self.college_sp = "44" # instance variable
clg = college()
# the varible which we create in constructor is called instance variable
```

```python
clg.college_name
```

```python
clg.college_address
```

```python
class Student:
    def __init__(self, name , roll_no,branch , classes):
        self.name = name
        self.branch = branch
        self.roll_no = roll_no
        self.classes = classes
    
    def display(self):
        print("Your name is:-",self.name)
        print("your branch is",self.branch)
        print("Your roll no.",self.roll_no)
        print("Your class is:-",self.classes)
st = Student(name = "garima",branch = "data science",roll_no = 18,classes = "second year")
st.display()
# we use this method to pass values 
```

```python
# constructor in multilevel

class Student:
    def __init__(self,name,age,hieght):
        self.name = name
        self.age = age
        self.hieght = hieght
class Teacher(Student):
    def __init__(self,name,age,hieght,mentor):
        super().__init__(name,age,hieght)
        self.mentor = mentor
class HOD(Teacher):
    def __init__(self,name,age,hieght,mentor,address):
        super().__init__(name,age,hieght,mentor)
        self.address = address
    def display(self):
        print("Your name is:-",self.name)
        print("Your age is :-",self.age)
        print("your hieght:-",self.hieght)
        print("mentor name:-",self.mentor)
        print("your address:-",self.address)

hod = HOD(name = "garima",age = 19,address="jaipur",hieght=5.4,mentor="dev")
hod.display()
```
