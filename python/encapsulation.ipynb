Encapsulation in Python is one of the core concepts of Object-Oriented Programming (OOP). It means wrapping data (variables) and methods (functions) together into a single unit (class) and restricting direct access to some parts of the object.

🔑 Key Idea

Encapsulation helps:
* Protect data from accidental modification
* Control how data is accessed or changed
* Improve code security and maintainability

|Access Type  |	Syntax|	Meaning|
------------|--------------|-------------------------------------|
|Public|	self.name|	Accessible everywhere|
|Protected|	self._name|	Should not be accessed outside class (convention)
|Private|	self.__name| Strongly restricted (name mangling)

```python
class college:
    college_name = "SGI" # public variable
    college_garden = "yes" # public variable
    college_account = "12345678" #public variable
clg = college()
```

```python
clg.college_account
```

```python
class college:
    college_name = "SGI" # public variable
    college_garden = "yes" # public variable
    __college_account = "12345678" # private variable
# we denote any variable as private by adding two underscores before the variable
clg = college()
```

```python
 # clg.__college_account it will show error because the variable is private
```

```python
class college:
    college_name = "SGI" # public variable
    _college_garden = "yes" # protected variable
    college_account = "12345678" # public variable
# protected variable is denote by one underscore before the variable name
clg = college()
```

```python
clg._college_garden
```

```python
class college:
    college_name = "SGI" # public variable
    college_garden = "yes" # public variable
    __college_account = "12345678" # private variable
    def display(self): # self is object, keyword and parameter
        #return self.__college_account # after returning the value it will print it
        print(self.__college_account) # we can also use print in place of return
clg = college() 
clg.display()
```
