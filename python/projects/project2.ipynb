```python
# BMI US UNITS

class BMI:
    def __init__(self,):
        self.W= float(input("Enter your wieght in pound :-"))
        self.H = float(input("Enter your hieght in feet:-"))
        self.B = "b"
    def checkbmi(self):
        # to convert wieght of pound in kg we devide pound by 2.205
        self.W = self.W / 2.205
        print(f"your wieght is:- {self.W} in kg")

        # to convert hieght of feet in meter we devide feet by 3.281
        self.H = self.H / 3.281
        print(f"Your hieght is:- {self.H} in meter")
        
        self.B = self.W / self.H ** 2
        print(f"BMI:- {self.B} kg/m^2")
    def checkhealth(self):
        if self.B < 16:
            print("Severe Thinness")
        elif 16<self.B<17:
            print("Moderate Thinness")
        elif 17<self.B<18.5:
            print("Mild Thinness")
        elif 18.5<self.B<25:
            print("Normal")
        elif 25<self.B<30:
            print("Overwieght")
        elif 30<self.B<35:
            print("Obese Class I")
        elif 35<self.B<40:
            print("Obese Class II")
        else:
            print("Obese Class III")
b = BMI()
b.checkbmi()
b.checkhealth()






#OUTPUT:-
```

```python
# BMI Metric UNITS

class BMI:
    def __init__(self,):
        self.W= float(input("Enter your wieght in kg :-"))
        self.H = float(input("Enter your hieght in cm:-"))
        self.B = "b"
    def checkbmi(self):
        print(f"your wieght is:- {self.W} in kg")

        # to convert hieght of cm into meter we devide cm by 100
        self.H = self.H / 100
        print(f"Your hieght is:- {self.H} in meter")
        
        self.B = self.W / self.H ** 2
        print(f"BMI:- {self.B} kg/m^2")
    def checkhealth(self):
        if self.B < 16:
            print("Severe Thinness")
        elif 16<self.B<17:
            print("Moderate Thinness")
        elif 17<self.B<18.5:
            print("Mild Thinness")
        elif 18.5<self.B<25:
            print("Normal")
        elif 25<self.B<30:
            print("Overwieght")
        elif 30<self.B<35:
            print("Obese Class I")
        elif 35<self.B<40:
            print("Obese Class II")
        else:
            print("Obese Class III")
b = BMI()
b.checkbmi()
b.checkhealth()


# OUTPUT:-
```
