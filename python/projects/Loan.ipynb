```python
pay = sat["loan_approved"].value_counts()
plt.pie(pay,labels=pay.index,autopct='%1.1f%%',data=sat)
plt.title("percent of possibility")
plt.show()
```

```python
ratings = sat.groupby("loan_approved")[[
    "income",
    "credit_score",
    "loan_amount"
]].mean()

ratings.plot(kind="bar")
plt.title("Average Loan Ratings")
plt.ylabel("Average prediction")
plt.xticks(rotation=0)
plt.show()
```
