### 4. Create a class FactorialCalculator with a function to calculate the factorial of a number using a loop.

```python
class FactorialCalculator:
    
    def __init__(self, num):
        self.num = num

    def factorial(self):
        fact = 1

        for i in range(1, self.num + 1):
            fact = fact * i

        print("Factorial is:", fact)
a = int(input("Enter your number: "))

obj = FactorialCalculator(a)
obj.factorial()
```

### 5. Create a class PrimeNumberChecker with a function to check whether a number is prime or not. 

```python
class PrimeNumberCheck:
    def prime(self):
        for z in range(1,26):
            if z>1:
                for q in range(2,z):
                    if z%q==0:
                        break
                else:
                    print("this is prime",z)
c = PrimeNumberCheck()
c.prime()
```

### 8. Create a class Employee using a constructor (__init__) to store employee details and create a function to calculate yearly salary. 

```python
class Employee:
    def __init__(self,name,age,salary):
        self.name = name
        self.age = age
        self.monthly_salary = salary
        
    def Salary(self):
        self.yearly_salary = self.monthly_salary * 12
        return self.yearly_salary
    
    def show(self):
        print("your name is :- ",self.name)
        print("your age is:-",self.age)
        print("your monthly salary:-",self.monthly_salary)
        print("Your yearly salary is :- ",self.Salary())
emp = Employee(name = "garima",age = 19, salary = 23000)
emp.show()
```

#### Create a class Calculator with functions: 
○ add() ○ subtract() ○ multiply() ○ divide()

```python
class Calculator:
    def add(self):
        a = float(input("Enter a number"))
        b = float(input("input another number"))
        self.c = a + b
        print("sum=",self.c)
    def sub(self):
        d = float(input("Enter your number"))
        e = float(input("Enter another number"))
        self.f = d - e
        print("sub=",self.f)
    def mul(self):
        g = float(input("Enter a number"))
        h = float(input("Enter another number"))
        self.i = g * h
        print("Multiplication=",self.i)
    def devide(self):
        j = float(input("Enter a number"))
        k = float(input("Enter another number"))
        self.l = j / k
        print("Devide = ",self.l)
cal = Calculator()  
cal.add()
cal.sub()
cal.mul()
cal.devide()
```
