## Looping in python
A loop in Python is used to repeat a block of code multiple times.

Python mainly has 2 types of loops:

for loop
while loop

1. <span style = "color:yellow"> For Loop </span>

A for loop is used when you know how many times you want to repeat something.

2. <span style = "color:yellow"> While Loop </span>

A while loop runs as long as a condition is True.



#

```python
name = "garima"
for x in name :
    print(x)
```

```python
lst = [2,3,5,6,6]
for y in lst:
    print(y)
```

```python
for x in range(1,20):
    print(x)
```

```python
for x in range(2,22,2):
    print(x)
```

```python
dict1 = {"name": "garima",
         "roll no.":19,
         "address":"jaipur"
         }
print(dict1)
for x in dict1.items(): # keys() and values() will be same for it.
    print(x)
```

```python
tpl = (2,3,4,5,5,6,7,8,9)
for item in tpl:
    print(item)
```

```python
lst = [2,5,6,7,2,5]
for x in lst:
    if x == 7:
        continue # it is used to skip element which we pass
        # break is used to break the loop from element which we passed
    print(x)
```

```python
fruits = ["apple", "banana", "mango","grapes"]
for x in fruits:
    if x == "mango" :
        continue
    print(x)
```

```python
lst = [3,5,6,8,["hello","hII","wooohh"]]
print(lst[4])
print(lst[4][2])
```

```python
## While looping:
     #block of code 


x = 1
while x<= 5:
    print(x)
    x += 1
```

```python
x = 5
while x >= 1:
    print(x)
    x -= 1
```
