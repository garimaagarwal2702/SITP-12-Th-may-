## function 
**Fuction** fuction are a block of code which performs particular task.
fuction performs tasks when it is called.

*Funtion are of two type*
* Predefined function
* User defined function

**Predefined function** or **built in function**
* print()  |	  Displays output	   print("Hello")
* input()  |  Takes user input	   name = input()
* len()	    |  Finds length	       len("Python")
* type()	|  Shows data type	   type(10)
* int()	   |   Converts to integer  int("5")
* float()  |  Converts to float	   float("3.5")
* str()	  |     Converts to string   str(100)
* list()  |  Converts into list   list("abc")
* sum()	 |    Adds values	       sum([1,2,3])
* max()	  |    Largest value	       max(4,8,2)
* min()	  |    Smallest value	   min(4,8,2)
* sorted()	 Sorts data	       sorted([3,1,2])
* abs()	     Absolute value	   abs(-5)
* round() 	  Rounds number	       round(3.456)

```python
def even(lst):
    print(lst)
    for x in lst:
        if x % 2 == 0:
            print("even:-",x)
        else :
            print("odd:-",x)
lst1 = [2,3,4,5,6,7,2,2,1,7,8,9]
even(lst = lst1)
```

```python
def grade(a):
    print(a)
    for x in a:
        if x>=90 and x<=100:
            print("A+")
        elif x>=80 and x<=90:
            print("A")
        elif x>=70 and x<=80:
            print("B")
        elif x>=60 and x<=70:
            print("C")
        elif x>=40 and x<=60:
            print("D")
        else:
            print("F")
b = [23,45,67,46,57]
grade(a = b)

```

```python
def prime():
    for x in range(1,26):
        if x>1:
            for i in range(2,x):
              if x % i==0:
                break
            else:
              print(x)
prime()
```
