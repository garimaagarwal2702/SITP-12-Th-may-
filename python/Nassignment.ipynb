```python
# 1. Create a Numpy array of 50 random numbers and find 
# the 2nd largest and 3rd smallest element.

import numpy as np 
arr = np.random.randint(1,100,50)
print("this is all the random 50 numbers:- ", arr)
arr1 = np.sort(arr)
print("this is the sorted form of arr:- ", arr1)
print("the second largest element is:- ", arr1[-2])
print("the third smallest element is:- ", arr1[2])
```
