#### What is Pandas in Python?
pandas is a powerful Python library used for data analysis and data handling.
It helps us work with:
* Tables of data
* Excel files
* CSV files
* Large datasets
* Data cleaning and analysis

It is widely used in:
* Data Science
* Machine Learning
* Data Analysis
* AI projects
* Business reporting
* Why Do We Use Pandas?

Without Pandas, handling large data in Python becomes difficult.
Pandas makes it easy to:

#### 1. Store Data in Table Form

Example:
Name	Marks	City
Garima	89	Jaipur
Rahul	75	Delhi

Pandas stores this using a DataFrame.

#### 2. Read Files Easily

You can read:
CSV files
Excel files
JSON files

Example:
import pandas as pd
data = pd.read_csv("students.csv")
print(data)

#### 3. Clean Data

Pandas helps remove:
Empty values
Duplicate data
Wrong formatting

Example:
data.dropna()

#### 4. Analyze Data

You can calculate:
Sum
Average
Maximum
Minimum

Example:
print(data["Marks"].mean())

#### 5. Filter Data

Example:
print(data[data["Marks"] > 80])

This shows students with marks greater than 80.

### Main Objects in Pandas
#### 1. Series
A single column of data.

Example:
import pandas as pd
a = pd.Series([10, 20, 30])
print(a)

#### 2. DataFrame
A full table with rows and columns.

Example:
import pandas as pd
data = {
    "Name": ["Garima", "Rahul"],
    "Marks": [89, 75]
}
df = pd.DataFrame(data)
print(df)
