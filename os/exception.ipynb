```python
class InvalidAgeError(Exception):
    pass
try:
    age = 17
    if age<18:
        raise InvalidAgeError ("you are not eligible")
    print("your can vote")
except InvalidAgeError as e:
    print(e)
```

```python
class Insufficient(Exception):
    pass
try:
    bal = 5000
    a = int(input("enter withdrawal money"))
    if a > bal:
        raise Insufficient("balance is not sufficient")
    print("you can take out money")
except Insufficient as e:
    print(e)
```

```python
class PasswordLen(Exception):
    pass
try:
    password = 12345678
    a = input("Enter your password")
    if a!=password:
        raise PasswordLen("len of your password is longer")
    print("correct paasword")
except PasswordLen as e:
    print(e)
```

```python
class EmailCheck(Exception):
    pass
try:
    email = input("Enter your email")
    if "@" not in email:
        raise EmailCheck ("Email has not @",email)
    if "." not in email:  
        raise EmailCheck ("Email has not .",email)
    if isupper() in email:
        raise EmailCheck ("Email has not capital letter",email)
    if islower() in email:
        raise EmailCheck ("Email has not lower letter",email)
    print("valid email")
except EmailCheck as e:
    print(e)
```
