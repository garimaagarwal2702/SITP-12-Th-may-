```python
file=open('requirement.txt','w')
file.write("hii how are you ?")
file.close()
file=open("requirement.txt",'r')
file.read()
```

```python
file=open('poit.txt','w') # we can also create a file with the help of w
file.write("hello") # there was no existing file named poit but w created it and write in it
file.close()
file=open('poit.txt','r')
file.read()
```

```python
file = open('poit.txt','a')
file.write("File is bieng handled very cleverly by this methods.")
file.close()
```
